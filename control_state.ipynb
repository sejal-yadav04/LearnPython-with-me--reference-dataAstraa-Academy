#2. For Loops

```python
grades = [98, 76, 77, 81, 100, 96, 69, 75, 83, 83]
total=0
count=0
for marks in grades:
    count+=1
    if marks>100:
        print("ivalid marks")
        continue
    total+=marks
print("total grade=", total)
```

```python
# calculating average 
average= total/count
```

#3.While Loop

```python
grades = [98, 76, 77, 81, 100, 96, 69, 75, 83, 83]
grade_counter = 1
total = 0  # initialize total

while grade_counter <= len(grades):
    grade = grades[grade_counter - 1]

    if grade >= 100:
        print("Invalid Grades!! at", grade_counter)
        grade_counter += 1   # increment here too to avoid infinite loop
        continue
    else:
        total += grade

    grade_counter += 1   # increment counter after processing each grade

print("Average is", round(total / len(grades)))
```

#5. Dictionary Iteration


```python
my_dict = {
    'B1': {'Topic': 'Python', 'Status': 'Ongoing', 'students': ['Vinay', 'Deepika', 'Gaurav']},
    'B2': {'Topic': 'R', 'Status': 'Ongoing', 'students': ['Deepika', 'Gaurav', 'Manish']}
}

students=[]
for batch_id, batch_details in my_dict.items():
    if batch_details['Status']=='Ongoing':
        students.append(batch_details['students'])
print(set(students[0]).intersection(set(students[1])))
```
