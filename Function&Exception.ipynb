#Variable argument (*arg)

```python
def arithmetic_mean(x, *l):
    sum = x
    for i in l:
        sum += i
    return sum / (1.0 + len(l))
l = [23, 34, 56, 75, 78, 49, 6]
print(arithmetic_mean(10, *l))
```

```python
def details(**data):
    for k, v in data.items():
        print(f"{k} = {v}")
details(Name="Sejal", LastName="Yadav")  # Name = Vinay, LastName = Borhade
```

```python
from functools import partial
```

```python

def power(base, exponent):
    return base ** exponent

# Create a new function with exponent fixed to 2
square = partial(power, exponent=2)

print(square(5))  # 5^2 = 25
print(square(8))  # 8^2 = 64
```

```python
def recurse(k):
    if k > 0:
        result = k + recurse(k - 1)
        print(result)
    else:
        result = 0
    return result
recurse(3)
```

```python
l = [12, 14, 13, 56, "vinay", "14"]
div = 0
try:
    for i in l:
        print(i / div)
except ZeroDivisionError:
    print("you cannot divide by zero")
except TypeError:
    print("input should be a number")
```

```python
class MyError(Exception):
    pass
def divide(x, y):
    try:
        if x == 0 and y == 0:
            raise MyError
        result = x / y
        print("result is", result)
    except MyError:
        print("Are you nuts!")
    except ZeroDivisionError:
        print("**Invalid Denom** - division by zero!")
    finally:
        print("executing finally clause")
divide(0, 0)
```
